# Bias Variance Trade-Off

Bias - When the machine learning model cannot find a good pattern/relationship in the training data. Very low bias corresponds to Overfitting, and high Bias referes to Underfitting.

Variance - Difference between the error of the model on the training and the test data. Note that in the case of overfitting case, there will be high variance.

We should go for Low Bias and Low Variance
