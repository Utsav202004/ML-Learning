# Lasso Regression and Sparsity

What is Sparsity: In the case of lasso regression, If we increase lambda, a lot of coefficients corresponding to input columns becomes zero. This filetring of the input features is called as Sparsity. 

But how is lasso able to do this filtering while Ridge is not able to do so.

## For single feature and a single Output

The thing that is different in lasso is that, the term added to the loss function is sum of modulus of the values, which was sum of squares(L2 norm) in the case of Ridge regression. Now the loss function is not differentiable due to the mod, so we will need to make two different cases. Now the use of these two cases is what makes the difference

The lambda is in numerator in the case of lasso, while it is in denominator in Ridge, which can drive it close to zero but never equal to zero, while in the numerator it can make it zero.
