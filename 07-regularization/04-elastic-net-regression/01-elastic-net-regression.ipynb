# ElasticNet Regression/Regularisation

This is a combination of Ridge and Lasso regression.

We use Ridge in the case when we know that all the input columns are important and we do not need to remove some features, while in lasso, we know we need a subset of those input features.

Elastic net is done when we do not know beforehand if we there are unecessary input features

we add both L1 and L2 terms to the loss function.

a : Coefficient of the L2(Ridge) term and b: Coefficient of the L1(Lasso) term. Lambda : a(def= 0.5) + b(def = 0.5) and l1_ratio(paramter, def = 1) = (a)/(a+b)

This l1_ratio implies what ratio of a particular regression we want to apply.

We use Elatic net when there is multicolinearity in the dataset. Example: Height and weight.

## Lets work on some data

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge,Lasso, ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

## Simple Linear Regression

```python
reg = LinearRegression()
reg.fit(X_train, y_train)
y_pred = reg.predict(X_test)
r2_score(y_test, y_pred)
```

## Ridge Regression

```python
reg = Ridge(alpha=0.1)
reg.fit(X_train, y_train)
y_pred = reg.predict(X_test)
r2_score(y_test, y_pred)
```

## Lasso Regression

```python
reg = Lasso(alpha=0.01)
reg.fit(X_train, y_train)
y_pred =  reg.predict(X_test)
r2_score(y_test, y_pred)
```

## Elastic Net

```python
reg = ElasticNet(alpha=0.005, l1_ratio  = 0.9) # we are giving more weightage to 'a'->L2 norm->Ridge
reg.fit(X_train, y_train)
y_pred = reg.predict(X_test)
r2_score(y_test, y_pred)
```

There are two ways of using Elastic net - The elastic net from sklearn or using SGD regressor and using penalty paramater as elastic net. Same goes for l1 and l2 individually.
