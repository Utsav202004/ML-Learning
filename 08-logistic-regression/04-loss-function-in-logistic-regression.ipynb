Till now, we have worked on perceptron trick and then the use of Sigmoid function to improve it, but still it was not as accurate as Logistic regression, here Machine learning plays a role. We can use the concept of Loss function as we have used earlier in linear regression and make the prediction more accurate.

We use the concept of Maximum Likelihood for the same.

If we are using Maximum likelihood, lets say to comapre two models, we will find the probability of each of the point(according to the model) and multiply it and decide which model is better.

Now multiplying the probability of a lot of points leads to a very very small number as the number of data points increase, so we use a different method, we use the log summation of these probability. As the number inside the log here is between 0 and 1, the log value will be negative, so we add the negative of the log values here (we call these numbers Cross Entropy)

We need to minimise the sum of this Cross Entropy. (As the log(a) > log(b) when a < b and a and b are between 0 and 1)

Note that the probability is of the point itself and not the simple sigmoid probabilty we used earlier. So we need to come up with a formula which takes this into consideration as well

Loos Function - We do the summation of ( -ylog(y_hat) - (1 - y)*log(1 - y_hat) ), and divide by n -> this is called as Log-loss error or Binary Cross Entropy
