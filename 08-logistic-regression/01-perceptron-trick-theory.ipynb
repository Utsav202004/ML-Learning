# Logistic Regression

Pre-requisite to apply Logistic Regression: The data should be linearly separable or almost linearly separable. If completely non linear data, will not give good results.

General equation of the line in Logistic Regression: AX1 + BX2 + C = 0

The objective is to find the values of A, B, and C.

Points above a line is in the positive region of the line.

All the three coefficient impacts the line:

1) A: The line will rotate about the y intercept.
2) B: The line will rotate about the X intercept.
3) C: The line will shift up or down with the slope being the same.

## Perceptron Trick

We need to do multiple Epochs(1000 generally) and each epochs involve cross checking the trueness of the line based on the initial random values of A, B and C. Each time we change the coefficient if the line does not fit the reality for that random point

How does this transformation occurs:

1) Find a point which is on the wrong side of the line(lets say (1,2))
2) we need to subtract(when in positive region)/add(when in negative region) (1,2,1) from the coefficients of the line (1,2) comes from the point and we need to take 1 always
3) Let the line be 2x1 + 3x2 + 4 = 0, so coefficients are (2,3,4)
4) So the new line is (1-2)x1 + (2-3)x2 + (1-4) = 0 -> -x1 - x2 - 3 = 0

Now this can be a drastic change, so we use a learning rate(like 0.01), we subtract learning_rate*(old coefficient) from the coefficients of the line.

We can make a single algo to handle all the four possible cases.
